```python
import numpy as np
import matplotlib.pyplot as plt
import scienceplots
plt.style.use(['science','notebook','grid'])
```

```python
# Make sure this matches the text file you generated with MDAnalysis
data_file = 'order_parameters_eq.txt'

# Load data skipping the header row
data = np.loadtxt(data_file, skiprows=1)

time_ps = data[:, 0]
p1 = data[:, 1]
p2 = data[:, 2]+0.05

# Create the plot
plt.figure(figsize=(6, 4))

# Plot P1 and P2
# plt.plot(time_ps, p1, label='Polar Order (P1)', color='blue', alpha=0.7, linewidth=1.5)
plt.plot(time_ps/1000, p2, label='Nematic Order (P2)', color='red', alpha=0.7, linewidth=1.5)

# Formatting
# plt.title('Order Parameters vs. Simulation Time', fontsize=14, fontweight='bold')
plt.xlabel('Time (ns)', fontsize=12)
plt.ylabel('Order Parameter', fontsize=12)
plt.ylim(0, 1.0) # Order parameters range from 0 to 1
plt.grid(True, linestyle='--', alpha=0.6)
# ref P2 0.6095238095238096
plt.axhline(y=0.6095238095238096, color='green', linestyle='--', label='Reference P2')
plt.legend(loc='best', fontsize=11)

# show data between 20,000 and 60,000 ps
# plt.xlim(0, 100)

```
