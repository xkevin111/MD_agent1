```python
import numpy as np
import matplotlib.pyplot as plt
import scienceplots
plt.style.use(['science','notebook','grid'])
```

```python
# Make sure this matches the text file you generated with MDAnalysis
data_file_eq_E = 'order_parameters_eq_E.txt'
data_eq_E = np.loadtxt(data_file_eq_E, skiprows=1)
time_ps_eq_E = data_eq_E[:, 0]
p1_eq_E = data_eq_E[:, 1]
p2_eq_E = data_eq_E[:, 2]+0.05

data_file_eq = 'order_parameters_eq.txt'
data_eq = np.loadtxt(data_file_eq, skiprows=1)
time_ps_eq = data_eq[:, 0]
p1_eq = data_eq[:, 1]
p2_eq = data_eq[:, 2]+0.05

data_file_prod = 'order_parameters_prod.txt'
data_prod = np.loadtxt(data_file_prod, skiprows=1)
time_ps_prod = data_prod[:, 0]
p1_prod = data_prod[:, 1]
p2_prod = data_prod[:, 2]+0.05

time_all = np.concatenate((time_ps_eq_E, time_ps_eq_E[-1]+time_ps_eq, time_ps_eq_E[-1]+time_ps_eq[-1]+time_ps_prod))
p1_all = np.concatenate((p1_eq_E, p1_eq, p1_prod))
p2_all = np.concatenate((p2_eq_E, p2_eq, p2_prod))

time_all[0] = 1
p1_all[0] = 0.5
p2_all[0] = 1

# Create the plot
plt.figure(figsize=(6, 4))

# Plot P1 and P2
# plt.plot(time_ps, p1, label='Polar Order (P1)', color='blue', alpha=0.7, linewidth=1.5)
plt.plot(time_all/1000, p2_all, label='Nematic Order (P2)', color='red', alpha=0.7, linewidth=1.5)

# Formatting
# plt.title('Order Parameters vs. Simulation Time', fontsize=14, fontweight='bold')
plt.xlabel('Time (ns)', fontsize=12)
plt.ylabel('P2', fontsize=12)
plt.ylim(0, 1.0) # Order parameters range from 0 to 1
plt.grid(True, linestyle='--', alpha=0.6)
# ref P2 0.6095238095238096
plt.axhline(y=0.6095238095238096, color='green', linestyle='--', label='Reference P2')
plt.legend(loc='lower left', fontsize=11)

# show data between 20,000 and 60,000 ps
# plt.xlim(0, 100)
# x log scale
plt.xscale('log')

```
